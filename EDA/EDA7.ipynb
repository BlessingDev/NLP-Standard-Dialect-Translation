```python
def remove_standard(path_to_data, region, token):
    file_path = os.path.join(path_to_data, "{0}_dialect_{1}_integration.csv".format(region, token))
    
    df = pd.read_csv(file_path, index_col=0)
    
    dialect_df = df[df['방언'] != df["표준어"]]
    out_file_path = os.path.join(path_to_data, "{0}_dialect_{1}_dialect_integration.csv".format(region, token))
    dialect_df.to_csv(out_file_path, index=False)
```

```python
regions = ["chungcheong", "gangwon", "gyeongsang", "jeju", "jeonla"]
tokens = ["bpe", "형태소", "jamo", "SentencePiece"]

for r in regions:
    for t in tokens:
        remove_standard("../datas/{0}/".format(r), r, t)
        print("{0} {1} 방언 부분집합만으로 구성 완료".format(r, t))
```

```python
def calculate_sample_size(path_to_data, region):
    token = "jamo"
    
    file_path = os.path.join(path_to_data, "{0}_dialect_{1}_dialect_integration.csv".format(region, token))
    
    df = pd.read_csv(file_path, index_col=0)
    
    train_size = len(df[df["셋"] == "train"])
    valid_size = len(df[df["셋"] == "val"])
    test_size = len(df[df["셋"] == "test"])
    
    print("{0} 방언 셋 크기: train {2}, val {3}, test {4}".format(region, token, train_size, valid_size, test_size))
```

```python
def calculate_orginal_sample_size(path_to_data, region):
    token = "jamo"
    
    file_path = os.path.join(path_to_data, "{0}_dialect_{1}_integration.csv".format(region, token))
    
    df = pd.read_csv(file_path, index_col=0)
    
    train_size = len(df[df["셋"] == "train"])
    valid_size = len(df[df["셋"] == "val"])
    test_size = len(df[df["셋"] == "test"])
    
    print("{0} 방언 셋 크기: train {2}, val {3}, test {4}".format(region, token, train_size, valid_size, test_size))
```

```python
regions = ["chungcheong", "gangwon", "gyeongsang", "jeju", "jeonla"]

for r in regions:
    calculate_sample_size("../datas/{0}/".format(r), r)
```

```python
regions = ["chungcheong", "gangwon", "gyeongsang", "jeju", "jeonla"]

for r in regions:
    calculate_orginal_sample_size("../datas/{0}/".format(r), r)
```
